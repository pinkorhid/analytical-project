```
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
# 1. Прибираємо користувачів, у яких tvod == 1
#filtered_df = user_df[user_df['tvod'] != 1].copy()
user_df = pd.read_csv(r'C:\Users\pinko\Documents\dell files\desktop\analytical project\data\user_df_sample.csv')
# 2. Обираємо тільки числові колонки (виключаємо user_pseudo_id та текстові/категоріальні поля)
numeric_df = user_df.select_dtypes(include=['number'])

# Прибираємо константні колонки (де у всіх однакове значення), щоб не було NaN у матриці
numeric_df = numeric_df.loc[:, numeric_df.nunique() > 1]

# 3. Рахуємо кореляційну матрицю
corr_matrix = numeric_df.corr(method='pearson')

# Виводимо топ-15 ознак, що найбільше корелюють з кількістю покупок
if 'purchases_count' in corr_matrix.columns:
    print("\nТоп кореляцій з кількістю покупок (purchase_revenue):")
    print(corr_matrix['purchase_revenue'].sort_values(ascending=False).head(15))

# 4. Будуємо теплову карту (Heatmap)
plt.figure(figsize=(15, 10))
sns.heatmap(
    corr_matrix,
    cmap='coolwarm',
    vmin=-1, vmax=1, center=0,
    annot=False,
    linewidths=0.5
)
plt.title('Кореляційна матриця', fontsize=16)
plt.tight_layout()
plt.savefig(r'C:\Users\pinko\Documents\dell files\desktop\analytical project\images\corr_matrix.png', dpi=150, bbox_inches='tight')
plt.show()
```
